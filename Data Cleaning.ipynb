```python
# fill empty cells
df['Discount'] = df['Discount'].fillna(df['Discount'].median())
df['Ship_Mode'] = df['Ship_Mode'].fillna(df['Ship_Mode'].mode()[0])
df['Region'] = df['Region'].fillna('Unknown')
df["Order_ID"] = df["Order_ID"].fillna("ORD-UNKNOWN")
```

```python
df[df["Sales"] < 0]
#print("rows before:", rows_before)
print("rows after :", len(df))
print("empty cells left:", df.isnull().sum().sum())
```

```python
df = df.copy()

# Standardize Region column
df['Region'] = df['Region'].str.strip().str.title()

# Standardize Category column
df['Category'] = df['Category'].str.strip().str.title()

# Correct inconsistent category names
df['Category'] = df['Category'].replace({
    'Office Supply': 'Office Supplies',
    'Furnitures': 'Furniture'
})
```

```python
# remove rows with no date, and rows with a date after Dec 2025 (not possible)
df = df.dropna(subset=['Order_Date'])
df = df[df['Order_Date'] <= '2025-12-31']
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
